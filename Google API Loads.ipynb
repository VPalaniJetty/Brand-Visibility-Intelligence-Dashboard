```python
# Dont run this code as it calls API for many times inside the loop.
# Below code is Ran for once and results are stored in "C:\Users\ASUS\OneDrive\Ai-ML Data Science Guvi\Project 2 Power Bi\all_results.json"
# Below code is to store all results available with api instead of getting only shopping results as shown in the above code 

import serpapi

client = serpapi.Client(api_key="25f2851096448d2cfc27cfc03f736bf862455c60de97ac5d458067294064c4c3")

Shopping_List = ['smart watches','sound bars','Projectors','Smart TVs',
                'fridge','ear phones','Tablets','washing Mechines',
                'Camera','Gaming Laptop','smart phones']
all_results = []
for shoplist in Shopping_List:
    for start in [0, 40, 80]: 
        results = client.search({
        "engine": "google_shopping",
        "q": shoplist,
        "location": "India",
        "start": start
        })

        shopping_results = results
        all_results.append(shopping_results)

print(f"Total Results: {len(all_results)}")
```

```python
GoogleShopResults = []
for result in all_results:
    #print(result)
    keyword = result.get('search_parameters').get('q')
    for data in result.get('shopping_results'):
        #print(data)
        GoogleShopResults.append({
        'keyword':keyword,
        'title':data.get('title'),
        'price':data.get('price'),
        'rating':data.get('rating'),
        'reviews':data.get('reviews'),
        'platform':data.get('source'),
        'delivery':data.get('delivery'),
        'discounts':data.get('tag'),
        'position':data.get('position'),
        'data from':'Google'})
```

```python
# Plotting a boxplot of the 'price' column to visualize the distribution of product prices, identify outliers, and understand the central tendency and variability of prices in the dataset. 
# This visualization can help in making informed decisions in Power BI by highlighting price ranges and potential anomalies.
plt.boxplot(shopping_results['price']) 
plt.show()
```

```python
# This tells us the value below which 99% of the prices lie and helps in identifying outliers and understanding the distribution of product prices in the dataset.
q1 = shopping_results['price'].quantile(0.25)
q3 = shopping_results['price'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}, IQR: {iqr}")
print(f"Q1: {q1}, Q3: {q3}, IQR: {iqr}")
```

```python
outliers = shopping_results[shopping_results['price'] > upper_bound]
outliers
```

```python
# Checking the outliers which are more than 75% quartile to understand which products have unusually high prices, which could be due to premium features, 
# brand value, or data entry errors. This information is useful for analysis and visualization in Power BI to identify trends and anomalies in product pricing.
outliers[['title', 'brand', 'price']].sort_values('price', ascending=False) 
```

```python
# Plot histogram of prices to identify outliers and understand the distribution of product prices in the dataset. The histogram provides a visual representation of how prices are spread across different ranges, which can help in identifying trends and making data-driven decisions in Power BI.

plt.hist(shopping_results['price'][shopping_results['price'] >= upper_bound], bins=30) 
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Distribution of Product Prices')
plt.show()
```
